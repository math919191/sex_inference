```python
import seaborn

import pandas as pd
import numpy as np
simmap_file = '../ped-sim/refined_mf.simmap'

```

```python
from sex_inference_analysis import main

sibs_cousins_F_input30 = "../ped-sim/outputs/sex-inference/sibs-cousins-F-30.seg"
sibs_cousins_F_output30 = f'results/sibs_cousin_F_results30.txt'

sibs_cousins_M_input30 = "../ped-sim/outputs/sex-inference/sibs-cousins-M-30.seg"
sibs_cousins_M_output30 = f'results/sibs_cousin_M_results30.txt'


main(
    input_file=sibs_cousins_F_input30,
    output_file=sibs_cousins_F_output30,
    map_file=simmap_file,
    bim="",
    window=50,
    count_unknown_ibd2_co=True
)

main(
    input_file=sibs_cousins_M_input30,
    output_file=sibs_cousins_M_output30,
    map_file=simmap_file,
    bim="",
    window=50,
    count_unknown_ibd2_co=True
)

df_sibs_cousins_F_30 = pd.read_csv(sibs_cousins_F_output30, sep='\t')
df_sibs_cousins_M_30 = pd.read_csv(sibs_cousins_M_output30, sep='\t')

```

```python
import seaborn as sns
import matplotlib.pyplot as plt

set1 = df_sibs_cousins_F_30['LOD_all']
set2 = df_sibs_cousins_M_30['LOD_all']


df = pd.DataFrame({
    'value': np.concatenate([set1, set2]),
    'dataset': ['Set 1'] * len(set1) + ['Set 2'] * len(set2)
})

# Plot
sns.histplot(
    data=df, 
    x='value', 
    hue='dataset', 
    palette={'Set 1': 'salmon', 'Set 2': 'skyblue'},
    alpha=0.5,
    bins=30,
    kde=False
)

plt.show()

```
